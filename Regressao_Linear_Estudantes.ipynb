```python
import pandas as pd

url = "https://raw.githubusercontent.com/Sainathmudigonda/Student-Result-Analysis/main/student_exam_scores.csv"

df = pd.read_csv(url)

df.head()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Definir as variáveis
X = df[['hours_studied']]
y = df['exam_score']

# Criar e treinar o modelo
modelo = LinearRegression()
modelo.fit(X, y)

# Calcular as previsões
y_pred = modelo.predict(X)

# Exibir os resultados
print("Inclinação:", modelo.coef_[0])
print("Intercepto:", modelo.intercept_)
print("R²:", r2_score(y, y_pred))
```

```python
import matplotlib.pyplot as plt

plt.scatter(X['hours_studied'], y, alpha=0.5)
plt.plot(X['hours_studied'], y_pred, color='red')

plt.xlabel('Horas de estudo')
plt.ylabel('Nota da prova')
plt.title('Regressão Linear: Horas de Estudo x Nota da Prova')

plt.show()
```
